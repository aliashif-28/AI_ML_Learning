i have 4 litter jar and 3 litter jar 

Question:- 4 litter jar fill in 2 litters of water 
           [jug A = 2, jug B = n]

#### Step 1

|Inital State |jug A = 0,| jug B = 0|
|-|-|-|
|Goal State =| jug A = 2, |jug B = n|


#### Step 2 :- Rule

|Rule 1 |jug A full fill|
|-|-|
|Rule 2 |jug B full fill|
|Rule 3 |jug A Empty|
|Rule 4 |jug B Empty|
|Rule 5 |Transfer jug A to jug B|
|Rule 6 |Transfer jug B to jug A|
|Rule 7 |Transfer some water of jug A to jug B|
|Rule 8 |Transfer some water of jug B to jug A|

#### Step 3:- Solution

|index|initial state|Rule|
