# DOMAIN OF AI

1) Expert Systems
-> An expert systems are AI programs that simulate the decision-making ability of a human expert. They use a knowledge base and rules to solve complex problems in a specific domain. These systems are used in medical diagnosis, finance, and technical support. Example:- Dendral, Mycin, DXplain.

2) Robotics
-> Robotics is the field of AI that deals with designing and developing intelligent robots. These robots can perform tasks automatically by sensing their environment and making decisions. Robotics is used in manufacturing, healthcare, and space exploration.

3) Machine Learning
-> Machine Learning is a branch of AI that allows computers to learn from data and improve their performance without being explicitly programmed. Algorithms analyze patterns in data and make predictions or decisions. It is used in recommendation systems, spam filtering, and fraud detection.

4) Natural Language Processing
-> Natural Language Processing (NLP) enables computers to understand, interpret, and generate human language. It helps machines read text, understand speech, and respond in natural language. Applications include chatbots, language translation, and voice assistants. Example:- Alexa, Siri.

5) Fuzzy Logic
-> Fuzzy Logic is an AI technique that deals with uncertain or approximate reasoning instead of fixed true or false values. In classical logic, values are either 0 (false) or 1 (true), but in fuzzy logic values can be between 0 and 1. It helps machines make decisions similar to human thinking.

6) Neural Network/ Deep Learning
-> Deep learning is a type of machine learning based on artificial neural networks in which multiple layers of processing are used to extract higher level features from data.These techniques are used in image recognition, speech recognition, and self-driving cars.

# WATER JUG PROBLEM

- You are given two jugs, 4-ltr and 3-ltr.
- Neither jug has any measuring markings on it.
- A pump which has unlimited water which you can choose to fill the jug.
- How can you get exactly 2 ltr of water in a 4 tr jug?
- We will represent a state of the problem as a tuple(x,y).
- x = amount of water in the 4 ltr jug
- y = amount of water in the 3 ltr jug
1) Initial state : (0,0); 0<=x<=4, and 0<=x<=3.
2) Goal state : (2,n) for any value of n.
